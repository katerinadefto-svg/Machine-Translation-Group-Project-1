```python
import pandas as pd
```

Φόρτωση Δεδομένων

```python
greek_corpus = "ParaCrawl_subset.el-en.el"
en_corpus = "ParaCrawl_subset.el-en.en"
```

Αποθήκευση σε Λίστες

```python
lst_greek = []
lst_english = []
```

```python
with open (greek_corpus,"r", encoding="utf-8") as f:
  for line in f:
    lst_greek.append(line.strip())
```

```python
with open (en_corpus,"r", encoding="utf-8") as f1:
  for line in f1:
    lst_english.append(line.strip())
```

```python
len(lst_greek)
```

```python
len(lst_english)
```

```python
lst_english[0]
```

Δημιουργία Dataframe

```python
data = pd.DataFrame({"Greek sentences": lst_greek,
                     "English translation": lst_english})
```

Προσθήκη στηλών με το μήκος της κάθε πρότασης τόσο για τα ελληνικά όσο και για τα αγγλικά & Στατιστικά

```python
# Βρίσκουμε το μήκος σε επίπεδο χαρακτήρων (Αυτό, αν δεν μας χρησιμεύσει πουθενά, μπορούμε να το αφαιρέσουμε στο τέλος).
data['gr_len'] = data["Greek sentences"].str.len()
data['en_len'] = data["English translation"].str.len()
```

```python
# Βρίσκουμε το μήκος σε επίπεδο λέξεων.
data["Word Count_GR"] = data["Greek sentences"].str.split().str.len()
data["Word Count_EN"] = data["English translation"].str.split().str.len()
```

```python
data.head()
```

```python
data.describe()
```

Έλεγχος Διπλότυπων

```python
data = data.drop_duplicates(subset = ["Greek sentences", "English translation"])
```

```python
len(data) # Δεν υπάρχουν διπλότυπα ζεύγη όπου και η ελληνική πρόταση και η αγγλική μετάφραση να είναι πανομοιότυπες.
```

```python
data = data.drop_duplicates(subset = ["Greek sentences"])
```

Plot the histogram of the word distribution

```python
import matplotlib.pyplot as plt

data["Word Count_EN"].plot.hist(bins=20)

plt.title("Distribution of Sentence Lengths (English)")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")

plt.show()
```

```python
data["Word Count_GR"].plot.hist(bins=20)

plt.title("Distribution of Sentence Lengths (Greek)")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")

plt.show()
```

We see that the distribution of sentence lengths between the two languages are similar and the majority of them fall between a length of 0-25 words.

```python
short_sentences = data[data["Word Count_GR"] <= 25]
print(f"Sentences available in the range of 0-25 words: {len(short_sentences)}")
```

We see that in the range of 0-25 words we have enough data to sample our training corpus of 300000 sentences

```python
sampled_df = short_sentences.sample(n=300000, random_state=42)
```

```python
sampled_df.head()
```

```python
# Save to plain text files (one sentence per line, no headers or row numbers)
sampled_df['English translation'].to_csv('english_sentences.txt', index=False, header=False)
sampled_df['Greek sentences'].to_csv('greek_sentences.txt', index=False, header=False)
```
