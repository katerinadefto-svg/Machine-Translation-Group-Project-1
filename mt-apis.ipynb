# MT with Commercial APIS

```python
from google_translate import translate_sentences

with open("test.el", encoding="utf-8") as file:
    sentences = [line.strip() for line in file if line.strip()]

```

```python
batch_size = 50
translations = []

for i in range(0, len(sentences), batch_size):
    batch = sentences[i : i + batch_size]
    translations.extend(translate_sentences(batch))
```

```python
with open("google_translations.en", "w", encoding="utf-8") as output_file:
    output_file.write("\n".join(translations))
```

```python
import time
from deepl_translate import translate_sentences

batch_size = 30 # DeepL has a max of 50
translations = []

for i in range(0, len(sentences), batch_size):
    batch = sentences[i : i + batch_size]
    translations.extend(translate_sentences(batch))
    time.sleep(2)
```

```python
with open("deepl_translations.en", "w", encoding="utf-8") as output_file:
    output_file.write("\n".join(translations))
```

```python
import time
from azure_translate import translate_sentences

batch_size = 25
translations = []

for i in range(0, len(sentences), batch_size):
    batch = sentences[i : i + batch_size]
    translations.extend(translate_sentences(batch))
    time.sleep(10)  # Safe for the Free (F0) tier
```

```python
with open("azure_translations.en", "w", encoding="utf-8") as output_file:
    output_file.write("\n".join(translations))
```
