Αποθήκευση των προτάσεων των dataset σε αρχεία json

```python
from google.colab import drive
drive.mount("/content/drive")
```

```python
import json
import os
```

```python
filename = "Flores_ell" + ".json"
path = os.path.join(f"/content/drive/MyDrive/Μηχανική Μετάφραση_Εργασία 1", filename)

with open(path, "w", encoding="utf-8") as f:
    json.dump(list(ell_dataset["sentence"]), f, ensure_ascii=False, indent=4)
```

```python
filename = "Flores_eng" + ".json"
path = os.path.join(f"/content/drive/MyDrive/Μηχανική Μετάφραση_Εργασία 1", filename)

with open(path, "w", encoding="utf-8") as f:
    json.dump(list(eng_dataset["sentence"]), f, ensure_ascii=False, indent=4)
```
